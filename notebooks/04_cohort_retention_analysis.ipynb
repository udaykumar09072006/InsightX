# InsightX — 04: Monthly Acquisition Cohort Retention
### Tracking 13 Acquisition Cohorts across 12-Month Longitudinal Decay

```python
import json

with open('data/processed/cohort_retention.json', 'r') as f:
    cohort_data = json.load(f)

print(f"Identified Customer Base: {cohort_data['total_identified_customers']:,}")
print(f"Repeat Purchasing Customers: {cohort_data['repeat_customers']:,}")
print(f"Single-Order Customers: {cohort_data['single_purchase_customers']:,}")
print(f"Repeat Purchase Rate: {cohort_data['repeat_purchase_rate_pct']}%")

print('\n--- COHORT RETENTION DECAY (SAMPLE COHORTS) ---')
for c in cohort_data['cohorts'][:5]:
    pcts = [f"{p['retention_pct']:.1f}%" if p['retention_pct'] is not None else 'N/A' for p in c['periods'][:7]]
    print(f"Cohort {c['cohort']} (Size {c['initial_size']:>4}): {' -> '.join(pcts)}")
```

### Analytical Notes:
- **High Inherent Repeat Rate**: 65.58% of customers place 2 or more orders over time, typical for wholesale/gift re-orders.
- **Decay Curve**: The 2010-12 baseline cohort stabilizes around 35-40% monthly retention after initial month 1 attrition.
