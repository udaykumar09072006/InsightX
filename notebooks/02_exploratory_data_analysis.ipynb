# InsightX — 02: Exploratory Data Analysis (EDA)
### Investigating Sales Velocity, Seasonality, Geographic Reach, and Basket Size

```python
import json
import csv

with open('data/processed/kpi_summary.json', 'r') as f:
    kpi = json.load(f)

print(f"Gross Sales: £{kpi['total_gross_revenue']:,.2f}")
print(f"Total Completed Orders: {kpi['total_orders']:,}")
print(f"Total Units Sold: {kpi['total_units_sold']:,}")
print(f"Identified Customer Count: {kpi['unique_identified_customers']:,}")
print(f"Average Order Value: £{kpi['average_order_value']:.2f}")

print('\n--- MONTHLY REVENUE PROGRESSION ---')
for m in kpi['monthly_trends']:
    mom_str = f"{m['mom_growth_pct']:+.1f}%" if m['mom_growth_pct'] is not None else 'Baseline'
    print(f"{m['month']}: £{m['gross_revenue']:>12,.2f} | Orders: {m['orders']:>5} | MoM: {mom_str}")
```

### Business Observations:
- **Q4 Surge**: November 2011 is the peak trading month with gross revenue exceeding £1.50M (+32.1% MoM over October).
- **December 2011 Censoring**: December 2011 only captures transactions up to December 9, explaining the apparent month-end drop.
- **Geographic Concentration**: The United Kingdom generates over 82% of total transaction value, followed by the Netherlands, EIRE, Germany, and France.
