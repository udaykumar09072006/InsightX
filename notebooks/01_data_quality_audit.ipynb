# InsightX — 01: Data Quality Audit & Profiling
### Dataset: UCI Machine Learning Repository Online Retail (541,909 Transactions)

This notebook establishes the foundational data hygiene and quality audit for the UCI Online Retail dataset.
Every filtering decision is audited, documented, and reconciled against raw source records.

```python
import json
import csv
from pathlib import Path

# Load audit summary compiled during pipeline execution
with open('data/processed/data_audit_summary.json', 'r') as f:
    audit = json.load(f)

print('--- DATA QUALITY AUDIT RESULTS ---')
print(f"Total Raw Records: {audit['total_raw_rows']:,}")
print(f"Exact Duplicates Removed: {audit['exact_duplicates_removed']:,}")
print(f"Cancellations (Invoice starting with 'C'): {audit['cancellations_count']:,}")
print(f"Negative Quantity Non-Cancellations (Damaged/Adjustments): {audit['negative_quantity_non_cancellations']:,}")
print(f"Zero or Negative Unit Price Lines: {audit['zero_or_negative_unit_price']:,}")
print(f"Missing Customer IDs (Unregistered/Guest): {audit['missing_customer_id']:,} ({audit['missing_customer_id']/audit['total_raw_rows']*100:.2f}%)")
print(f"Valid Completed Transaction Lines: {audit['valid_completed_transactions']:,}")
print(f"Total Gross Sales: £{audit['total_gross_revenue']:,.2f}")
print(f"Total Cancellation / Refund Value: £{audit['total_cancellation_value']:,.2f}")
print(f"Net Sales: £{audit['net_revenue']:,.2f}")
```

### Key Audit Conclusions & Cleaning Policies:
1. **Missing Customer IDs**: 135,080 rows (~24.9%) lack a CustomerID. Rather than discarding 25% of business revenue, these are classified as `GUEST` transactions for product/macro sales, and segregated during customer-level RFM and Cohort analysis.
2. **Cancellations**: Invoices prefixed with 'C' indicate refunds/cancellations totaling £461,844.20. Gross sales and net revenue are explicitly tracked.
3. **Duplicates**: 5,268 records are exact duplicates across all columns and safely purged.
