# InsightX — 03: Customer RFM Segmentation
### Recency, Frequency, and Monetary (RFM) Matrix for 4,372 Identified Customers

```python
import json

with open('data/processed/rfm_segments_summary.json', 'r') as f:
    rfm = json.load(f)

print(f"Reference Date: {rfm['reference_date']}")
print(f"Total Customers Segmented: {rfm['total_customers']:,}")
print(f"Total Identified Revenue: £{rfm['total_revenue']:,.2f}\n")

print(f"{'Segment':<28} | {'Cust Count':<10} | {'Cust %':<8} | {'Revenue (£)':<14} | {'Rev %':<8} | {'Avg Recency':<12}")
print('-' * 95)
for s in rfm['segments']:
    print(f"{s['segment']:<28} | {s['customer_count']:<10} | {s['customer_pct']:<8}% | £{s['total_revenue']:<12,.2f} | {s['revenue_pct']:<8}% | {s['avg_recency_days']:<12} days")
```

### Strategic Takeaways:
- **Champions Dominance**: 977 Champions (22.3% of customers) generate £5.64M (63.3% of all identified spend), with an average order frequency of 8.9 orders and recency of 15.6 days.
- **At-Risk Interventions**: 621 customers (£742K historical spend) have drifted beyond 180 days since their last transaction. Targeted reactivation campaigns are high ROI opportunities.
